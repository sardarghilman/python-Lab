```python
# Variables and basic data types
name = "GHILMAN"  # str
age = 21      # int
height = 5.9  # float
is_student = True  # bool

print(name, age, height, is_student)
print(type(name), type(age), type(height), type(is_student))
```

Operators

```python
a, b = 15, 7
print("Add:", a + b)
print("Sub:", a - b)
print("Mul:", a * b)
print("Div:", a / b)
print("Floor Div:", a // b)
print("Modulus:", a % b)
print("Power:", a ** b)
print("Comparison:", a > 2, a == b)
print("Logical:", (a > 5) and (b < 5))
```

If


```python
age = 20

if age >= 18:
    print("You are eligible to vote")
```

If Else


```python
num = 10

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

If Elif Else


```python
marks = 85

if marks >= 90:
    print("A+")
elif marks >= 80:
    print("A")
elif marks >= 70:
    print("B")
else:
    print("Fail")
```

Comparison Operators


```python
a = 10
b = 20

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

Logical Operators

```python
print(a > 5 and a < 20)
print(a > 15 or a < 20)
print(not(a > 5))
```

Break and Continue


```python
for i in range(1, 6):
    if i == 3:
        break
    print(i)
```

Tuples

```python
numbers = (10, 20, 30, 40)

print(numbers)
print(numbers[0])
print(len(numbers))
print(numbers.count(10))
print(numbers.index(20))
```

Dictionary


```python
student = {
    "name": "GHILMAN",
    "age": 22,
    "marks": 85
}

print(student["name"])
print(student["age"])

student["marks"] = 90

print(student)
```
