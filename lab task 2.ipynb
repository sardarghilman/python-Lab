```python

import random

REWARD_CLEAN = 10
COST_MOVE = 1
STEPS = 20
NUM_STATES = 10
STATE_SEED = 2026


# VACUUM ENVIRONMENT

class VacuumEnvironment:

    def __init__(self, status, agent_location):
        self.status = dict(status)
        self.agent_location = agent_location

    def percept(self):
        return (
            self.agent_location,
            self.status[self.agent_location]
        )

    def execute(self, action):

        if action == "Suck":
            self.status[self.agent_location] = 0

        elif action == "Right":
            self.agent_location = "B"

        elif action == "Left":
            self.agent_location = "A"


# SIMPLE REFLEX AGENT

def simple_reflex_agent(percept, memory):

    location, dirty = percept

    if dirty:
        return "Suck"

    if location == "A":
        return "Right"

    return "Left"


# RANDOM AGENT

def random_agent(percept, memory):

    return random.choice(["Left", "Right", "Suck"])


# RUN AGENT

def run_agent(agent_fn, status, location, steps=STEPS):

    env = VacuumEnvironment(status, location)

    memory = {"A": None, "B": None}
    score = 0

    for _ in range(steps):

        percept = env.percept()

        action = agent_fn(percept, memory)

        if action == "Suck":

            if env.status[env.agent_location] == 1:
                score += REWARD_CLEAN

            env.execute(action)

        elif action in ("Left", "Right"):

            score -= COST_MOVE

            env.execute(action)

    return score


# GENERATE RANDOM INITIAL STATES

def make_initial_states(n=NUM_STATES, seed=STATE_SEED):

    rng = random.Random(seed)

    states = []

    for _ in range(n):

        status = {
            "A": rng.randint(0, 1),
            "B": rng.randint(0, 1)
        }

        location = rng.choice(["A", "B"])

        states.append((status, location))

    return states


# EVALUATE AGENTS

def evaluate(agent_fn, name, states, repeats=1):

    scores = []

    for status, location in states:

        for _ in range(repeats):

            scores.append(
                run_agent(agent_fn, status, location)
            )

    average = sum(scores) / len(scores)

    print(
        f"{name:<20}: Average Score = {average:6.2f} "
        f"(Runs = {len(scores)})"
    )

    return average


# TASK 3 COMPARISON

def task3_compare():

    print("=" * 60)
    print("TASK 3: VACUUM WORLD AGENT COMPARISON")
    print("=" * 60)

    states = make_initial_states()

    print("\nInitial States:")

    for i, (status, location) in enumerate(states, start=1):

        print(f"State {i}: {status} | Agent at {location}")

    print("\nAgent Performance:\n")

    random.seed(7)

    simple_avg = evaluate(
        simple_reflex_agent,
        "Simple Reflex Agent",
        states
    )

    random_avg = evaluate(
        random_agent,
        "Random Agent",
        states,
        repeats=50
    )

    print("\nObservations:")

    print(
        f"Simple Reflex Agent Average Score: {simple_avg:.2f}"
    )

    print(
        f"Random Agent Average Score: {random_avg:.2f}"
    )

    if simple_avg > random_avg:
        print("Simple Reflex Agent performs better.")
    elif random_avg > simple_avg:
        print("Random Agent performs better.")
    else:
        print("Both agents have equal performance.")


# MAIN PROGRAM

task3_compare()
```
